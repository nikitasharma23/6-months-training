```python
#elif
a=10
b=10
if a>b:
    print("a is greater than b")
elif a==b:
    print("a is equal to b")
```

```python
day=int(input("Enter a day in number form:"))
if day==1:
    print("Monday")
elif day==2:
    print("Tuesday")    
elif day==3:
    print("Wednesday")  
elif day==4:
    print("Thursday")
elif day==5:
    print("Friday")
elif day==6:
    print("Saturday")
elif day==7:
    print("Sunday")
else:
    print("Invalid day")
        
```

```python
#else
a=10
b=20
if b<a:
    print("b is less than a")
else:
    print("b is greater than a")
```

```python
number=int(input("Enter a number:"))
if number%2==0:
    print("The number is even")
else:
    print("The number is odd")
```

```python
#nested statements
a=int(input("Enter a number:"))
if a>10:
    print("Number is greater than 10")
    if a>20:
        print("Number is greater than 20")
    else:
        print("But number is not greater than 20")
else:
    print("Number is less than 10 and 20")
```
