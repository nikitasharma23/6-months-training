```python
dict1={
    "name":"nikita",
    "age":21,
    "course":"AI/ML"
}
print(dict1)
```

```python
#defining the dictionary with dict() method
dict2=dict(name="nikita",age=21,course="AI/ML")
print(dict2)
```

```python
x=dict2["name"]
print(x)
```

```python
print(dict2["age"])
```

```python
#accessing the values using get() method
print(dict2.get("name"))    
```

```python
#accessing the keys using keys() method
print(dict2.keys())
```

```python
#accessing the values using values() method
print(dict2.values())
```

```python
#accessing the items using items() method
print(dict2.items())
```

```python
#changing the values
dict2["age"]=22
print(dict2)
```

```python
#changing values using update() method
dict2.update({"age":23})    
print(dict2)
```

```python
#adding the key values in dict2
dict2["section"]="A"
print(dict2)
```

```python
#adding the key values in dict2 using the update 
dict2.update({"color":"white"})
print(dict2)
```

```python
#deleting the key with values uaing pop method
dict2.pop("color")
print(dict2)
```

```python
#clear all the key value pairs using clear() method
dict2.clear()
```
