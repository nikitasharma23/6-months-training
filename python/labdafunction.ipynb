```python
#It is a anonymous function(def)
#It takes a single expression and it returns the result automatically of that expression
#Reason for using lambda function : =1 for writing the short code
#2. For creating the temporary function 
#3. map,filter and sorted
#syntax of lambda function:= lambda arguments: expression
```

```python
#parameter 
add=lambda a,b:a+b
print(add(10,20))
```

```python
#having zero parameter
greet=lambda: "Hello Python"
print(greet)
```

```python
#Find the length of string
length=lambda text:len(text)
print(length("Python"))
```

```python
#Normal function
def square(n):
    return n*n
print(square(5))
```

```python
#map in lambda function: map(function,iterable)
numbers=[1,2,3,4]
result=list(map(lambda x:x*2,numbers))
print(result)
```

```python
marks=[60,70,80,90]
updated_marks=list(map(lambda m:m+5,marks))
print(updated_marks)
```

```python
#check even 
numbers=[1,2,3,4,5,6,7,8]
result=list(filter(lambda x:x%2==0,numbers))
print(result)
```

```python
#sorted(iterable,key=None,revese=False)
numbers=[50,10,40,20,30]
result=sorted(numbers,reverse=True)
print(result)
```
