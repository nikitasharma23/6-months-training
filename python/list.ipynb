```python
#lists are mutable
list[1]="blackcurrent"
print(list)
```

```python
list[1:3]=["blackcurrent","Watermelon"]
print(list)
```

```python
#to add elemnents in list at any position we use insert() method
list.insert(2,"grapes")
print(list)
```

```python
#use remove to remove elements from list
list.remove("grapes")
print(list)
```

```python
#use pop to remove from specific index
list.pop(3)
print(list)
```

```python
#del method for deleting the specifix index element
del list[0]
print(list)
```

```python
#clear is used to remove all elements from the list
list.clear()
print(list)
```

```python
#using append for adding elements append is used to only add at the end
list.append("Orange")
print(list)
```

```python
list=["Apple","Banana","Mango","cherry"]
print(list)
print(len(list))
```

```python
print(list[0])#accessing element 
print(list[-1])#negative indexing
print(list[1:3])#slicing
print(list[:3])#indexing from start to 3
print(list[1:])
```
