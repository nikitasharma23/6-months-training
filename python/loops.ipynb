```python
#with the while loop we can execute a set of statements as long as a condition is true.
#NOTE - Dont't forget to add the increment in i or any variable, if it don't contain then loop will continue forever.
i=1
while i<6:
    print(i)
    i+=1
```

```python
#break statements
#using the break statement we can stop the loop even if the while condition is true.

i=1
while i<6:
    print(i)
    if i==4:
        break
    i+=1
```

```python
#continue statement
#we can stop the current function to the code and continue with next while condition is true
i=2
while i<12:
    i+=1
    if i==4:
        continue
    print(i)
```

```python
i=2
while i<12:
    i+=1
    if i<10:
        continue
    print(i)
```

```python
#we can use the for loop in a list,tuple,dictionary,set and a string also
name=["Raju","Karan","Amer","sonu"]
print(type(name))
for y in name:
    print(y)
```

```python
#looping with a string
for z in "raju":
    print(z)
```

```python
#use the break statements is for loop
for y in name:
    print(y)
    if y=="Amer":
      break
```
