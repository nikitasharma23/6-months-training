```python
#set is built in datatype
#set doesn't contain duplicate values
#set items are unchangeable,but we can also remove items and add new items
#set are unordered
#set are the values inside the middle braces{}
```

```python
sets={"apple","banana","cherry","watermelon",1,2,3,4}
print(sets)
print(type(sets))
```

```python
sets1={"apple","banana","cherry",True,1,2,0,False}
print(sets1)
print(len(sets1))
```

```python
#accessing elements with the help of keywords
print("banana" in sets1)
print("banana" not in sets1)
```

```python
#adding elements in sets
sets1.add("orange")
print(sets1)
```

```python
#removing elements from sets
sets1.remove("banana")
print(sets1)
```

```python
#removing elements from sets using discard -> it doesnot throw an error if the element is not present in the set
sets1.discard("cherry")
print(sets1)
```

```python
# sets1.remove("watermelon")
# print(sets1)

sets1.discard("watermelon")
print(sets1)
```

```python
sets1.pop()
print(sets1)
```

```python
#clear
sets1.clear()
print(sets1)
```
