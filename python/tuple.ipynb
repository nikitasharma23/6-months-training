```python
tuple=('apple', 'banana', 'cherry')
print(tuple)
```

```python
#tuples are unchangeable/immutable
#tuples are ordered
#tuples can store duplicate values
```

```python
#len() function returns the number of items in tuple
print(len(tuple))
```

```python
#access the tuple items
print(tuple[1])#positive indexing
print(tuple[-1])#negative indexing
print(tuple[0:3])
print(tuple[::])
print(tuple[:3])
print(tuple[1:3])
print(tuple[1:])
```

```python
tuple2 = ("apple", "banana", "cherry")

# 1. Convert the tuple to a list
y = list(tuple2)

# 2. Append the new item to the list
y.append("watermelon")

# 3. Convert the list back into a tuple
tuple2=tuple(y)

print(tuple2)

```
